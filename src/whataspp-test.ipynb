```python
import json
from pathlib import Path
from parsers.whatsapp import WhatsAppParser

# 1. Input your exact display name as it appears in the WhatsApp chat export
MY_WHATSAPP_NAME = "NY" # 
WHATSAPP_EXPORT_FILE = "" # Update to your actual file path

# 2. Parse the WhatsApp chat log
print("Parsing WhatsApp messages...")
parser = WhatsAppParser(my_name=MY_WHATSAPP_NAME)
samples = parser.parse(WHATSAPP_EXPORT_FILE)

# 3. Define the destination folder expected by the notebook
raw_dir = Path.home() / "style-llm" / "data" / "raw"
raw_dir.mkdir(parents=True, exist_ok=True)
output_file = raw_dir / "whatsapp_samples.jsonl"

# 4. Save to JSON Lines format (.jsonl)
valid_samples = 0
with open(output_file, "w", encoding="utf-8") as f:
    for sample in samples:
        # Optional: you can filter out very short strings or media text here like "<Media omitted>"
        if "<Media omitted>" not in sample["text"] and len(sample["text"].split()) >= 3:
            f.write(json.dumps(sample) + "\n")
            valid_samples += 1

print(f"Successfully saved {valid_samples} training samples to: {output_file}")
```
